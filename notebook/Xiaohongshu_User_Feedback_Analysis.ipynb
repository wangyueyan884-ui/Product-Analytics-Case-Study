# Product Insights

## Key User Pain Points

### 1. Recommendation Experience

Users highly value personalized recommendations, but some users report repeated or less relevant content.

**Product Opportunity:**
Improve recommendation diversity and personalization algorithms.

---

### 2. Search Experience

Search is one of the most frequently mentioned topics.

**Product Opportunity:**
Optimize search ranking and improve content discovery efficiency.

---

### 3. Advertisement Experience

Some users feel that excessive advertisements affect browsing experience.

**Product Opportunity:**
Improve ad-user balance through better ad frequency control.

---

## Product Priority

Based on user feedback frequency:

1. Improve recommendation quality
2. Optimize search experience
3. Balance advertisement exposure
4. Enhance content quality

```python
# User pain points

category_counts = (
    df["category"]
    .value_counts()
    .sort_values()
)

plt.figure(figsize=(8,5))

plt.barh(
    category_counts.index,
    category_counts.values
)

plt.xlabel("Number of Reviews")
plt.ylabel("Category")
plt.title("Top User Feedback Categories")

plt.show()
```

```python
plt.savefig(
    "../visualization/sentiment_distribution.png",
    dpi=300,
    bbox_inches="tight"
)
```

```python
# Sentiment distribution visualization

sentiment_counts = df["sentiment"].value_counts()

plt.figure(figsize=(6,4))

plt.pie(
    sentiment_counts.values,
    labels=sentiment_counts.index,
    autopct="%1.1f%%"
)

plt.title("User Sentiment Distribution")

plt.show()
```

```python
plt.savefig("../visualization/rating_distribution.png",
            dpi=300,
            bbox_inches="tight")
```

```python
# Rating distribution

rating_counts = df["rating"].value_counts().sort_index()

plt.figure(figsize=(6,4))

plt.bar(
    rating_counts.index,
    rating_counts.values
)

plt.xlabel("Rating")
plt.ylabel("Number of Reviews")
plt.title("User Rating Distribution")

plt.show()
```
